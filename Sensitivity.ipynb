```python
import matplotlib.pyplot as plt
from tensorboard.backend.event_processing import event_accumulator
```

```python
scalar_tags = [
    "dbar_loss/beta_10.000000",
    "free_energy/beta_10.000000",
    "dbar_loss/beta_100.000000",
    "free_energy/beta_100.000000",
    "dbar_loss/beta_1000.000000",
    "free_energy/beta_1000.000000",
    "dbar_loss/beta_10000.000000",
    "free_energy/beta_10000.000000",
    "metrics/final_distortion",
]
```

```python
import os
import random
SEED = 1
random.seed(SEED)
rng = random.Random(SEED)

log_dir = "C:\\Users\\salar\\OneDrive\\Documents\\GitHub\\AutonomyAwareClustering\\runs\\2\\"

tag = "dbar_loss/beta_10.000000"


plt.figure(figsize=(8, 6))


# Find all event files in the parent directory
parent_dir = os.path.dirname(log_dir)
event_dirs = [
    d for d in os.listdir(parent_dir) if os.path.isdir(os.path.join(parent_dir, d))
]

num_unique_colors = len(event_dirs)
import distinctipy
colors = distinctipy.get_colors(num_unique_colors)

for event_dir in event_dirs:
    event_path = os.path.join(parent_dir, event_dir)
    ea_file = event_accumulator.EventAccumulator(
        event_path, size_guidance={"scalars": 0}
    )
    ea_file.Reload()
    index_to_plot = 5
    # Check if the tag exists in this event file
    if tag in scalar_tags:
        events = ea_file.Scalars(tag)
        steps = [e.step for e in events]
        values = [e.value for e in events]
        color = colors[event_dirs.index(event_dir)]
        plt.plot(steps[index_to_plot:], values[index_to_plot:], 
            linewidth=1.5, marker='.', markevery=10, color=color, alpha=0.7)

plt.xlabel("Step",fontsize=18)
plt.ylabel("$L(\\theta)$",fontsize=18)
# change tick font size
plt.xticks(fontsize=16)
plt.yticks(fontsize=16)
# plt.title(f"Loss $L(\\theta)$ over Training Steps")
# plt.legend()
# plt.grid(True)
plt.tight_layout()
plt.savefig(f"dbar_loss_2_beta_10.png", dpi=400)
plt.show()
```

```python
import os
import random
SEED = 1
random.seed(SEED)
rng = random.Random(SEED)

log_dir = "C:\\Users\\salar\\OneDrive\\Documents\\GitHub\\AutonomyAwareClustering\\runs\\1\\"

tag = "dbar_loss/beta_10000.000000"


plt.figure(figsize=(8, 6))


# Find all event files in the parent directory
parent_dir = os.path.dirname(log_dir)
event_dirs = [
    d for d in os.listdir(parent_dir) if os.path.isdir(os.path.join(parent_dir, d))
]

num_unique_colors = len(event_dirs)
import distinctipy
from mpl_toolkits.axes_grid1.inset_locator import inset_axes
colors = distinctipy.get_colors(num_unique_colors)

for event_dir in event_dirs:
    event_path = os.path.join(parent_dir, event_dir)
    ea_file = event_accumulator.EventAccumulator(
        event_path, size_guidance={"scalars": 0}
    )
    ea_file.Reload()
    index_to_plot = 0
    # Check if the tag exists in this event file
    if tag in scalar_tags:
        events = ea_file.Scalars(tag)
        steps = [e.step for e in events]
        values = [e.value for e in events]
        color = colors[event_dirs.index(event_dir)]
        plt.plot(steps[index_to_plot:], values[index_to_plot:], label = event_dir,
            linewidth=1.5, marker='.', markevery=10, color=color, alpha=0.8)

plt.xlabel("Step",fontsize=18)
plt.ylabel("$L(\\theta)$",fontsize=18)
# change tick font size
plt.xticks(fontsize=16)
plt.yticks(fontsize=16)
# Create inset plot in top right corner for steps >= 50
axins = inset_axes(plt.gca(), width="60%", height="60%", loc='upper right')

for event_dir in event_dirs:
    event_path = os.path.join(parent_dir, event_dir)
    ea_file = event_accumulator.EventAccumulator(
        event_path, size_guidance={"scalars": 0}
    )
    ea_file.Reload()
    if tag in scalar_tags:
        events = ea_file.Scalars(tag)
        steps = [e.step for e in events]
        values = [e.value for e in events]
        # Filter for steps >= 50
        filtered_steps = [s for s, v in zip(steps, values) if s >= 50]
        filtered_values = [v for s, v in zip(steps, values) if s >= 50]
        color = colors[event_dirs.index(event_dir)]
        axins.plot(filtered_steps, filtered_values, 
                  linewidth=1.0, marker='.', markevery=5, color=color, alpha=0.8)

axins.set_xlabel("Step", fontsize=16)
axins.set_ylabel("$L(\\theta)$", fontsize=16)
axins.tick_params(labelsize=14)


# plt.ylim(0, 100)
# plt.title(f"Loss $L(\\theta)$ over Training Steps")
# plt.legend()
# plt.grid(True)
# plt.tight_layout()
plt.savefig(f"dbar_loss_1_beta_10000.png", dpi=400)
plt.show()
```
